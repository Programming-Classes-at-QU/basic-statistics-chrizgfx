```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

df= pd.read_csv('hulu.csv')

# numeric and categorical colums
numCols = df.select_dtypes(include=['number']).columns
catCols = df.select_dtypes(include=['object', 'bool']).columns

# stats for numeric columns
numStats = df[numCols].describe().T
numStats['median'] = df[numCols].median()
numStats['mode'] = df[numCols].mode().iloc[0]
numStats['range'] = numStats['max'] - numStats['min']
numStats['std_dev'] = df[numCols].std()
catStats = df[catCols].nunique()

missingValues = df.isnull().sum()

# kurt and skew
var1, var2 = "show/episodes_count", "show/clips_count"
skewKurtosis = {
    var1: {"skew": skew(df[var1].dropna()), "kurtosis": kurtosis(df[var1].dropna())},
    var2: {"skew": skew(df[var2].dropna()), "kurtosis": kurtosis(df[var2].dropna())},
}

plt.figure(figsize=(10,5))
sns.histplot(df[var1].dropna(), bins=30, kde=True)
plt.title(f'histogram of {var1}')
plt.show()

# genre vs episodes count
filtered_df = df[['show/genre', var1]].dropna()
top_genres = filtered_df['show/genre'].value_counts().nlargest(6).index  # top 6
filtered_df = filtered_df[filtered_df['show/genre'].isin(top_genres)]

plt.figure(figsize=(12,6))
sns.boxplot(x="show/genre", y=var1, data=filtered_df)
plt.title(f'boxplot of {var1} by genre')
plt.show()

print("Numeric Statistics:")
print(numStats)

print("\nCategorical Statistics (Number of Unique Values):")
print(catStats)

print("\nMissing values in each column:")
print(missingValues)

print("\nSkewness and kurtosis for episode and clip count:")
print(skewKurtosis)
```
